```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill
import re 
# Load the Excel file
file_path = "C:/Users/muhammad.asad/OneDrive - YouGov Services Limited/Desktop/DataProcess Automation/CarMax/Final to client (script)/18 - 24 Apr/US26560581_CARMAX_2024-2025_Week of April 18 - 24,2025_Internal.xlsx"  # Replace with your file path
data = pd.read_excel(file_path, na_filter=False)
```

```python
# List of columns to delete
columns_to_delete = [
    "inputzipdma_recode", "OE Count 1.1", "OE Count 1.2", "OE Count 1.3", "OE Count 2.1", "OE Count 2.2", "OE Count 2.3",
    "Count of Low Frequency", "Content", "period", "identity", "cookieid", "ip_address_current",
    "ip_address_previous", "Missing_Words_Count"
]

# Drop the specified columns
data_cleaned = data.drop(columns=columns_to_delete, errors='ignore')
```

```python
# Rename the first column to RecordNo and fill it with ascending numbers where there is data
if not data_cleaned.empty:
    data_cleaned.rename(columns={data_cleaned.columns[0]: "RecordNo"}, inplace=True)
    data_cleaned["RecordNo"] = range(1, len(data_cleaned) + 1)
```

```python
# Rename specific columns
column_rename_map = {
    "birthyr": "Birth year",
    "bixdemo_ethnicity": "Ethnicity",
    "bixdemo_gender": "Gender",
    "categorytest": "Category",
    "bixdemo_income2": "Income",
    "bixdemo_kids": "Kids under 18",
    "pdl_Carmax2022_online_Q6": "inputzip"
}
data_cleaned.rename(columns=column_rename_map, inplace=True)
```

```python
# List of columns to relabel
columns_to_relabel = [
    "Carmax2022_online_Q2_1", "Carmax2022_online_Q2_3", "Carmax2022_online_Q2_4", "Carmax2022_online_Q2_5", 
    "Carmax2022_online_Q2_6", "Carmax2022_online_Q2_7", "Carmax2022_online_Q2_8", "Carmax2022_online_Q2_10", 
    "Carmax2022_online_Q2_11", "Carmax2022_online_Q2_12", "Carmax2022_online_Q2_99", 
    "Carmax2022_online_Q4_1", "Carmax2022_online_Q4_2", "Carmax2022_online_Q4_3", "Carmax2022_online_Q4_4", 
    "Carmax2022_online_Q5_1", "Carmax2022_online_Q5_3", "Carmax2022_online_Q5_4", "Carmax2022_online_Q5_5", 
    "Carmax2022_online_Q5_6", "Carmax2022_online_Q5_7", "Carmax2022_online_Q5_8", "Carmax2022_online_Q5_10", 
    "Carmax2022_online_Q5_9"
]

# Relabel cells in the specified columns
def relabel_cells(value):
    if value == "selected":
        return "Yes"
    elif value == "not selected":
        return "No"
    return value

for column in columns_to_relabel:
    data_cleaned[column] = data_cleaned[column].map(relabel_cells)
```

```python
# Relabel cells in the Category column
def relabel_category(value):
    if value == "internet":
        return "Online Sector"
    elif value == "retail_gasoline":
        return "Retail Auto gas sector"
    return value

data_cleaned["Category"] = data_cleaned["Category"].map(relabel_category)

# Save the cleaned data to a new Excel file
output_file_path = "US26560581_CARMAX_2024-2025_Week of April 18 - 24, 2025.xlsx"
data_cleaned.to_excel(output_file_path, index=False)
```
